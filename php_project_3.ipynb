```python
from IPython.display import HTML

html_content = """
<!DOCTYPE html>
<html>
<head>
    <title>Welcome to My Website</title>
    <style>
        body {
            margin: 0;
            font-family: Arial, sans-serif;
            text-align: center;
            background-color: #f2f2f2;
        }

        header {
            background-color: #4CAF50;
            color: white;
            padding: 15px;
        }

        .slider {
            position: relative;
            max-width: 800px;
            margin: 30px auto;
            overflow: hidden;
            border-radius: 10px;
            box-shadow: 0 0 10px gray;
        }

        .slides {
            display: none;
        }

        img {
            width: 100%;
            height: 400px;
        }

        .prev, .next {
            cursor: pointer;
            position: absolute;
            top: 50%;
            padding: 10px;
            color: white;
            font-weight: bold;
            font-size: 20px;
            background-color: rgba(0,0,0,0.5);
            border: none;
            transform: translateY(-50%);
        }

        .prev { left: 0; }
        .next { right: 0; }

        footer {
            background-color: #333;
            color: white;
            padding: 10px;
            margin-top: 20px;
        }
    </style>
</head>
<body>

<header>
    <h1>Welcome To My  Website</h1>
</header>

<div class="slider">
    <div class="slides">
        <img src="https://picsum.photos/id/1015/800/400">
    </div>

    <div class="slides">
        <img src="https://picsum.photos/id/1016/800/400">
    </div>

    <div class="slides">
        <img src="https://picsum.photos/id/1018/800/400">
    </div>

    <div class="slides">
        <img src="https://picsum.photos/id/1020/800/400">
    </div>

    <div class="slides">
        <img src="https://picsum.photos/id/1024/800/400">
    </div>

    <button class="prev" onclick="changeSlide(-1)">❮</button>
    <button class="next" onclick="changeSlide(1)">❯</button>
</div>

<footer>
    <p>© 2026 Simple Website | Created by Jagadeeswaran S</p>
</footer>

<script>
    let slideIndex = 0;
    showSlides(slideIndex);

    function changeSlide(n) {
        showSlides(slideIndex += n);
    }

    function showSlides(n) {
        let slides = document.getElementsByClassName("slides");
        if (n >= slides.length) slideIndex = 0;
        if (n < 0) slideIndex = slides.length - 1;
        for (let i = 0; i < slides.length; i++) {
            slides[i].style.display = "none";
        }
        slides[slideIndex].style.display = "block";
    }
</script>

</body>
</html>
"""
HTML(html_content)
```
